```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
sigma = 1.0
mu = 0.0

# Grid
x = np.linspace(-5, 5, 400)
f = (1/(np.sqrt(2*np.pi)*sigma)) * np.exp(-((x-mu)**2)/(2*sigma**2))

# Plot
plt.figure(figsize=(7,5))
plt.plot(x, f, color="orange")
plt.fill_between(x, f, 0, color='orange', alpha=0.5)
plt.xlabel("v")
plt.ylabel("Probability Density")
plt.title("1D Gaussian Velocity Distribution")

# Add equation label
equation = r"$f(v) = \frac{1}{\sqrt{2\pi}\sigma}\;\exp\left(-\frac{(v-\mu)^2}{2\sigma^2}\right)$"
plt.text(0.05, 0.9, equation, transform=plt.gca().transAxes, fontsize=13)

plt.grid(False)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Physical constants
kB = 1.380649e-23        # Boltzmann constant (J/K)
T = 333.0                # Temperature (K)
m = 87 * 1.66053906660e-27   # Mass of Rb-87 atom (kg)

# Standard deviation for 1D velocity distribution
sigma = np.sqrt(kB * T / m)
mu = 0.0

# Velocity grid (±6σ covers essentially all probability)
v_max = 6 * sigma
v = np.linspace(-v_max, v_max, 2000)

# Maxwell-Boltzmann 1D Gaussian distribution
f = (1/(np.sqrt(2*np.pi)*sigma)) * np.exp(-((v - mu)**2)/(2*sigma**2))

# Plot
plt.figure(figsize=(7,5))
plt.plot(v, f, color="orange")
plt.fill_between(v, f, 0, color='orange', alpha=0.5)

# Plot vertical bars
for z in range(0, 2000, 125):
    plt.plot([z-1000, z-1000], [0, f[z]], color='skyblue', linewidth=2)

# Axis labels with units
plt.xlabel("Velocity (m/s)")
plt.ylabel("Probability Density")
plt.title("1D Maxwell–Boltzmann Velocity Distribution for Rb-87 at 333 K")

# Gaussian equation label
equation = r"$f(v) = \frac{1}{\sqrt{2\pi}\sigma}\exp\left(-\frac{(v-\mu)^2}{2\sigma^2}\right)$"
plt.text(0.05, 0.9, equation, transform=plt.gca().transAxes, fontsize=13)

# Set ticks in m/s
# tick_range = np.arange(-v_max, v_max + 1, sigma)
plt.xticks([-1000, -750, -500, -250, 0, 250, 500, 750, 1000])

plt.grid(False)
plt.show()
```
